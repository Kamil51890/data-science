```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(1, 2, figsize = (9, 4))

ax[0].scatter(df['MedHouseVal'], df['Latitude'], s = 8, color = 'green', label = 'Latitude', alpha = 0.5)
ax[0].set_title('Распределение по штроте')
ax[0].set_xlabel('Цена дома')
ax[0].set_ylabel('Широта')
ax[0].legend()

ax[1].scatter(df['MedHouseVal'], df['Longitude'], s = 8, color = 'crimson', alpha = 0.5, label = 'Longitude')
ax[1].set_title('Распределение по долготе')
ax[1].set_xlabel('Цена дома')
ax[1].set_ylabel('Долгота')
ax[1].legend()
```

```python
df.plot.scatter(x = 'Longitude', y = 'Latitude', c = 'MedHouseVal')
```

```python
median_income = df['MedInc'].median()

low_income = df[df['MedInc'] <= median_income]

high_income = df[df['MedInc'] > median_income]

print(f'Низкий доход: {low_income['MedHouseVal'].median()}')
print(f'Высокий доход: {high_income['MedHouseVal'].median()} ')
```

```python
west = df[df['Longitude'] <= df['Longitude'].median()]
east = df[df['Longitude'] > df['Longitude'].median()]

print(f'Западное побережье: \n{west['MedHouseVal'].agg(['mean', 'median'])}\n')
print(f'Восточное побережье: \n{east['MedHouseVal'].agg(['mean', 'median'])}')
```
